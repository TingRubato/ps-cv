### AI-Assisted MIDI Arrangement Script

This script demonstrates how to generate a 32-bar Progressive House arrangement using `midiutil`. 
It includes logic for:
- Setting tempo and tracks
- Defining a Chord Progression (F Minor)
- Generating a Kick drum pattern
- Creating an Off-beat Bass (Simulated Sidechain)
- Generating a randomized Lead Arpeggio

```python
from midiutil import MIDIFile
import random

# Song Settings
tempo = 126
num_tracks = 3
duration = 32  # bars

# Create MIDI Object
mf = MIDIFile(num_tracks)

# Set Tempo
for i in range(num_tracks):
    mf.addTempo(i, 0, tempo)

# Root notes for bass: Fm - Db - Ab - Eb
bass_roots = [29, 29, 29, 29, 37, 37, 37, 37, 44, 44, 44, 44, 39, 39, 39, 39]

# 1. Kick Drum (4/4)
for bar in range(duration):
    for beat in range(4):
        # Track 0, Channel 9 (Percussion), Note 36 (C1), Time, Duration, Velocity
        mf.addNote(0, 9, 36, bar * 4 + beat, 1, 110) 

# 2. Off-beat Bass (Simulated Sidechain)
for bar in range(duration):
    root = bass_roots[bar % 16]
    for beat in [0.5, 1.5, 2.5, 3.5]:
        if not (8 <= bar < 16): # Breakdown silence
            vel = 100 + random.randint(-5, 5)
            mf.addNote(1, 0, root, bar * 4 + beat, 0.5, vel)

# 3. Lead Arpeggio
arp_notes = [65, 68, 72] # Fm Triad
for bar in range(duration):
    if bar >= 8:
        for sixteenth in range(16):
            pitch = random.choice(arp_notes)
            time = bar * 4 + (sixteenth * 0.25)
            # Humanize velocity
            vel = 80 + random.randint(-15, 15)
            mf.addNote(2, 1, pitch, time, 0.25, vel)

with open("ai_arrangement.mid", "wb") as outf:
    mf.writeFile(outf)
    print("MIDI file 'ai_arrangement.mid' generated successfully.")
```
